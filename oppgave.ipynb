## Gjennomsnittlig vekstfart

Den gjennomsnittlige vekstfarten til $f$ fra $x = a$ til $x = b$ er stigningstallet til linjen (sekanten) mellom punktene:

$$\frac{\Delta y}{\Delta x} = \frac{f(b) - f(a)}{b - a}$$

## Fra sekant til tangent

Hva skjer hvis vi lar $b$ komme nærmere og nærmere $a$? Vi skriver $b = a + h$ og lar $h$ bli liten:

$$\frac{f(a + h) - f(a)}{h}$$

Tabellen under viser dette for $f(x) = x^2$ i $a = 3$.

```python
def f(x):
    return x**2

a = 3
for h in [1, 0.1, 0.01, 0.001, 0.0001]:
    sekant = (f(a + h) - f(a)) / h
    print(f"h = {h:<7} stigningstall = {sekant}")
```

Stigningstallet nærmer seg 6. Dette kalles den **momentane vekstfarten** eller den **deriverte** i $x = 3$, og vi skriver $f'(3) = 6$. Grafisk er det stigningstallet til tangenten.

```python
import numpy as np
import matplotlib.pyplot as plt

xs = np.linspace(0, 5, 200)
plt.plot(xs, f(xs), "k", label="$f(x) = x^2$")
for h in [2, 1, 0.5]:
    s = (f(a + h) - f(a)) / h
    plt.plot(xs, s*(xs - a) + f(a), "--", label=f"sekant, h = {h}")
plt.plot(xs, 6*(xs - a) + f(a), "r", label="tangent")
plt.ylim(-5, 25)
plt.legend()
plt.grid(True)
plt.show()
```

## Derivasjonsregler for polynomer

| Funksjon | Derivert |
| --- | --- |
| $c$ (konstant) | $0$ |
| $x^n$ | $n\,x^{n-1}$ |
| $c \cdot f(x)$ | $c \cdot f'(x)$ |
| $f(x) + g(x)$ | $f'(x) + g'(x)$ |

Eksempel: $f(x) = 4x^3 - 2x + 7$ gir $f'(x) = 12x^2 - 2$.

Sympy kan kontrollere svarene dine med `sp.diff`.

```python
import sympy as sp
from IPython.display import display
x = sp.symbols('x')

display(sp.diff(4*x**3 - 2*x + 7, x))
```
